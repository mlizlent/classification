```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid1(x):
    return 1 / (1 + np.exp(-(0.5 * x + 3)))

def sigmoid2(x):
    return 1 / (1 + np.exp(-(5 * x + 11)))

x = np.linspace(-15, 10, 400)
```

```python
plt.figure(figsize=(9, 5))
plt.plot(x, sigmoid(x), label=r"sigmoid(x) = $\frac{1}{1 + e^{-x}}$", linewidth=2)
plt.plot(x, sigmoid1(x), label=r"sigmoid1(x) = $\frac{1}{1 + e^{-(0.5x + 3)}}$", linewidth=2)
plt.plot(x, sigmoid2(x), label=r"sigmoid2(x) = $\frac{1}{1 + e^{-(5x + 11)}}$", linewidth=2)
plt.axhline(y=0.5, color="red", linestyle="--", label="probability = 0.5")

plt.xlabel("x")
plt.ylabel("Probability")
plt.title("Comparison of Sigmoids")
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
