```python
import numpy as np
from sklearn.model_selection import train_test_split

X = np.arange(1, 21).reshape(10, -1)
y = np.zeros(10)
y[7:] = 1

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False
)

print("X_train:\n", X_train)
print("\ny_train:\n", y_train)
print("\nX_test:\n", X_test)
print("\ny_test:\n", y_test)

print(f"\nProportion of class 1 in train set: {np.mean(y_train):.4f}")
print(f"Proportion of class 1 in test set:  {np.mean(y_test):.4f}")
```

```python
X_large = np.arange(1, 201).reshape(100, -1)
y_large = np.zeros(100)
y_large[70:] = 1

X_train_s, X_test_s, y_train_s, y_test_s = train_test_split(
    X_large, y_large, test_size=0.2, stratify=y_large, random_state=42
)

print(f"Proportion of class 1 in train set: {np.mean(y_train_s):.4f}")
print(f"Proportion of class 1 in test set:  {np.mean(y_test_s):.4f}")
```
